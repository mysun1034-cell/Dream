# Ch.02-01 NumPy 표 시각화

**목표:** `[[25, 150], ...]` 가 **리스트처럼 보이지만** `np.array` 후에는 **(샘플 × feature) 표**가 됨을 **눈으로** 확인.

1. 리스트 vs ndarray — **같은 숫자, 다른 타입**
2. **표(table)** + **shape** `(3, 2)`
3. **행=생선, 열=feature** 색칠
4. **인덱싱** `X[0]`, `X[:, 0]` 하이라이트
5. scatter — 표의 두 열을 x,y 축으로


```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.colors import Normalize
import matplotlib.cm as cm

plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False
```

## 1) 입력 — 리스트의 리스트 (겉모양은 리스트)

```python
data_list = [
    [25, 150],
    [28, 200],
    [30, 180],
]
labels = ["빙어", "도미", "도미"]
feature_names = ["길이 (cm)", "무게 (g)"]

print("type:", type(data_list))
print("data_list:", data_list)
```

## 2) NumPy 표로 변환

```python
X = np.array(data_list)

print("type:", type(X))
print("X:\n", X)
print("shape:", X.shape, "→ (샘플 3, feature 2)")
print("dtype:", X.dtype)
```

## 3) 표 + shape 한눈에 (히트맵 + 숫자)

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4.2))

# --- 왼쪽: 숫자 표 ---
ax0 = axes[0]
ax0.axis("off")
ax0.set_title("NumPy 표 X  —  shape (3, 2)", fontsize=13, fontweight="bold", pad=12)

cell_text = [[str(v) for v in row] for row in X]
col_labels = feature_names
row_labels = [f"생선 {i} ({labels[i]})" for i in range(len(labels))]

table = ax0.table(
    cellText=cell_text,
    colLabels=col_labels,
    rowLabels=row_labels,
    loc="center",
    cellLoc="center",
)
table.auto_set_font_size(False)
table.set_fontsize(11)
table.scale(1.2, 1.8)
for (r, c), cell in table.get_celld().items():
    if r == 0:
        cell.set_facecolor("#4472C4")
        cell.set_text_props(color="white", fontweight="bold")
    elif c == -1:
        cell.set_facecolor("#D9E2F3")
        cell.set_text_props(fontweight="bold")
    else:
        cell.set_facecolor("#FFF2CC" if r % 2 else "#FFE699")

# --- 오른쪽: 히트맵 (값 크기 = 색) ---
ax1 = axes[1]
im = ax1.imshow(X, cmap="YlOrRd", aspect="auto")
ax1.set_xticks([0, 1])
ax1.set_xticklabels(feature_names)
ax1.set_yticks([0, 1, 2])
ax1.set_yticklabels(row_labels)
ax1.set_title("같은 데이터 — 색 = 숫자 크기", fontsize=13, fontweight="bold")
for i in range(X.shape[0]):
    for j in range(X.shape[1]):
        ax1.text(j, i, str(X[i, j]), ha="center", va="center", fontsize=14, fontweight="bold")
plt.colorbar(im, ax=ax1, fraction=0.046, pad=0.04, label="값")

plt.tight_layout()
plt.show()
```

## 4) 행·열 의미 — **행=샘플(생선), 열=feature**

```python
fig, ax = plt.subplots(figsize=(8, 4.5))
ax.set_xlim(0, 10)
ax.set_ylim(0, 6)
ax.axis("off")
ax.set_title("shape (3, 2) = 행 3개 × 열 2개", fontsize=14, fontweight="bold")

colors_row = ["#FFD966", "#A9D18E", "#A9D18E"]
for i in range(3):
    rect = mpatches.FancyBboxPatch(
        (1, 4.2 - i * 1.3), 8, 1.0,
        boxstyle="round,pad=0.02", linewidth=2, edgecolor="#333",
        facecolor=colors_row[i], alpha=0.9,
    )
    ax.add_patch(rect)
    ax.text(0.3, 4.7 - i * 1.3, f"행 {i}", fontsize=11, fontweight="bold", va="center")
    ax.text(2.0, 4.85 - i * 1.3, f"X[{i}] = {list(X[i])}", fontsize=11)
    ax.text(2.0, 4.55 - i * 1.3, f"= {labels[i]} 한 마리", fontsize=10, color="#444")

ax.annotate("", xy=(9.5, 5.0), xytext=(9.5, 1.5),
            arrowprops=dict(arrowstyle="->", color="#C00000", lw=2))
ax.text(9.7, 3.2, "axis=0\n(행 방향)", fontsize=10, color="#C00000", rotation=90, va="center")

ax.annotate("", xy=(2, 0.6), xytext=(8, 0.6),
            arrowprops=dict(arrowstyle="->", color="#0070C0", lw=2))
ax.text(4.5, 0.2, "axis=1  (열 = feature: 길이, 무게)", fontsize=10, color="#0070C0", ha="center")

plt.tight_layout()
plt.show()
```

## 5) 인덱싱 하이라이트 — `X[0]` vs `X[:, 0]`

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 3.8))

def draw_grid(ax, title, highlight_cells, subtitle):
    ax.set_title(title, fontsize=12, fontweight="bold")
    ax.set_xticks([0, 1])
    ax.set_xticklabels(feature_names)
    ax.set_yticks([0, 1, 2])
    ax.set_yticklabels([f"행{i}" for i in range(3)])
    for i in range(3):
        for j in range(2):
            face = "#FF6B6B" if (i, j) in highlight_cells else "#E8E8E8"
            ax.add_patch(mpatches.Rectangle((j - 0.45, 2.45 - i), 0.9, 0.9,
                                            facecolor=face, edgecolor="#333", lw=2))
            ax.text(j, 2 - i, str(X[i, j]), ha="center", va="center", fontsize=16, fontweight="bold")
    ax.set_xlim(-0.7, 1.7)
    ax.set_ylim(-0.3, 3.2)
    ax.text(0.5, -0.15, subtitle, transform=ax.transAxes, ha="center", fontsize=10, color="#555")

draw_grid(axes[0], "X[0]  — 첫 번째 생선 (한 행)", {(0, 0), (0, 1)}, f"= {X[0]}  →  길이+무게 한 세트")
draw_grid(axes[1], "X[:, 0]  — 모든 생선의 '길이'만", {(0, 0), (1, 0), (2, 0)}, f"= {X[:, 0]}  →  열 전체")

plt.tight_layout()
plt.show()
```

## 6) 표 → scatter (Ch.1 연결)

표의 **열 0 = x축**, **열 1 = y축**

```python
fig, ax = plt.subplots(figsize=(6, 5))
colors = ["#5B9BD5" if lb == "빙어" else "#ED7D31" for lb in labels]
ax.scatter(X[:, 0], X[:, 1], s=200, c=colors, edgecolors="white", linewidths=2, zorder=3)
for i in range(len(X)):
    ax.annotate(f"생선{i}\n{labels[i]}", (X[i, 0], X[i, 1]),
                textcoords="offset points", xytext=(8, 8), fontsize=10)
ax.set_xlabel(feature_names[0], fontsize=12)
ax.set_ylabel(feature_names[1], fontsize=12)
ax.set_title("NumPy 표 X의 두 열 = 2D 그래프", fontsize=13, fontweight="bold")
ax.grid(True, alpha=0.3)
ax.legend(handles=[
    mpatches.Patch(color="#5B9BD5", label="빙어"),
    mpatches.Patch(color="#ED7D31", label="도미"),
], loc="upper left")
plt.tight_layout()
plt.show()
```

## 7) 리스트 ×2 vs NumPy ×2 (왜 타입이 중요한지)

```python
row = [25, 150]
row_np = np.array([25, 150])

print("리스트 [25,150] * 2  →", row * 2)
print("NumPy  array * 2     →", row_np * 2)
print()
print("리스트 = 반복 붙이기 / NumPy = 각 숫자에 2 곱하기")
```
